```python
from langchain_openai import ChatOpenAI
```

```python
def get_runpod_url(model: str) -> str:
    return "https://y4vum0waas8o96-8000.proxy.runpod.net"
```

```python
MODEL_MAP = {
        "gpt-oss-20b": "openai/gpt-oss-20b",
        "qwen3-8b": "Qwen/Qwen3-8B",
        "gemma-3-12b-it": "google/gemma-3-12b-it",
        "gemma4": "google/gemma-4-26B-A4B-it"
    }

class SLLMModel(ChatOpenAI):
    

    def __init__(self, model_name: str, **kwargs):
        model = MODEL_MAP.get(model_name)

        if not model:
            raise ValueError(f"Model not found: {model_name}")
        
        base_url = get_runpod_url(model)

        if not base_url:
            raise ValueError(f"Runpod URL not found for model: {model}")

        kwargs['base_url'] = f"{base_url}/v1"
        kwargs['api_key'] = "EMPTY"

        if model == "Qwen/Qwen3-8B":
            kwargs['extra_body'] = {
                "chat_template_kwargs": {"enable_thinking": False}
            }

        super().__init__(model=model, **kwargs)
```

```python
if "gemma4" in MODEL_MAP:
    print("gemma4 mapping")
    client = SLLMModel(model_name="gemma4") 
```

```python
for chunk in client.stream(
    input = [
        {"role": "user", "content": "간단한 로또 추천기 파이선 코드를 작성해줘"}
    ]
):
    print(chunk.content)
```
